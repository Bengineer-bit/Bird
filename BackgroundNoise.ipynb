```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from scipy.signal import resample
from IPython.display import Audio

folder_path = r'Bird_sound'
file_1 = "SummerOutdoors.wav"
file_2 = "call1 (1).wav"

file_path_1 = os.path.join(folder_path, file_1)
file_path_2 = os.path.join(folder_path, file_2)

# Load the audio files
SR_1, z_1 = wavfile.read(file_path_1)
SR_2, z_2 = wavfile.read(file_path_2)

# Ensure both are mono
if len(z_1.shape) > 1:
    z_1 = z_1[:, 0]  # Take the first channel if stereo
if len(z_2.shape) > 1:
    z_2 = z_2[:, 0]

# Resample if sampling rates are different
if SR_1 != SR_2:
    if SR_1 > SR_2:
        z_2 = resample(z_2, int(len(z_2) * SR_1 / SR_2))
        SR_2 = SR_1
    else:
        z_1 = resample(z_1, int(len(z_1) * SR_2 / SR_1))
        SR_1 = SR_2

# Play original audio files
display(Audio(z_1, rate=SR_1))
display(Audio(z_2, rate=SR_2))

# Trim both signals to the same length
min_length = min(len(z_1), len(z_2))
z_1 = z_1[:min_length]
z_2 = z_2[:min_length]

# Combine audio and normalize to avoid clipping
combined_audio = z_1 + z_2
combined_audio = combined_audio / np.max(np.abs(combined_audio))  # Normalization

# Convert back to int16 for proper playback
combined_audio = np.int16(combined_audio * 32767)

# Play combined audio
display(Audio(combined_audio, rate=SR_1))
```
